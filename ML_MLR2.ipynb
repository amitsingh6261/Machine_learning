```python
import pandas as pd
df=pd.read_csv('/content/multilinear_regression_data_500.csv')
df
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
#data
df['Smoker'] = df['Smoker'].replace({'Yes': 1, 'No': 0})
x=df[['Age','BMI','Children','Smoker']]
y=df['Charges ($)']

#create and train and test model
model=LinearRegression()

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2, random_state=42)
model.fit(x_train,y_train)

# take input
Age=float(input("enter the age"))
BMI=float(input("enter the BMI"))
Children=int(input("enter no of children"))
Smoker=int(input("enter Smoker(yes=1,no=0)"))

user_input=[[Age,BMI,Children,Smoker]]
prediction=model.predict(user_input)

print("Predicted Charges($) :",prediction[0])

#finr value
print('intercept', model.intercept_)
print('slop',model.coef_[0])
print('R2 score',model.score(x,y))

#create graph
fig = plt.figure()

ax = fig.add_subplot(111, projection='3d')

# Original data
ax.scatter(
    df['Age'],
    df['BMI'],
    df['Charges ($)']
)

# User prediction
ax.scatter(
    Age,
    BMI,
    prediction[0],
    color='red',
    s=200
)

ax.set_xlabel('Age')
ax.set_ylabel('BMI')
ax.set_zlabel('Charges ($)')

plt.show()
```
