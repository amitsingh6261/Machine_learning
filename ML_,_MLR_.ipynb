```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
#data
data = {
   'Experience' :[1,2,3,4,5],
  'Education' : [1,1,2,2,3],
   'Salary': [4,5,7,8,10]
}

df = pd.DataFrame(data)
```

```python
# independent variablr
x=df[['Experience','Education']]

# dependent variable
y=df['Salary']
```

```python
# create model
model=LinearRegression()
```

```python
#train model
model.fit(x,y)
```

```python
#cofficients
print("Intercept", model.intercept_)
print("Cofficients", model.coef_)
```

```python
#prediction
new_employee=[[6,3]]
predicted_salary=model.predict(new_employee)
print("pridicted salary: ", predicted_salary[0])
```

```python
# graph
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection='3d')

ax.scatter(
  df['Experience'] ,
  df['Education'],
  df['Salary'],
s=80
)
ax.set_xlabel('Experience(years)')
ax.set_ylabel('Education Level')
ax.set_zlabel('Salary(Lakh)')

ax.set_title('Experience+Education+Salary')
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
data ={
    'Area' :[800,1000,1200,1400,1600,1800,2000,2200,2500,2800],
    'Location_score' :[5,6,7,8,6,9,8,10,7,9],
    'Distance' :[10,8,7,5,12,4,6,3,10,5],
    'Price':[42,58,72,91,82,115,118,145,125,155]
}

df=pd.DataFrame(data)
df

# step2: independent Variable
x=df[['Area','Location_score','Distance']]
#Dependent Variable
y=df['Price']

# step 3 Creat Model
model=LinearRegression()
# step 5: Train Model
model.fit(x,y)

# step 6: Coefficents
print("Intercept:", model.intercept_)
print("Area Coefficients:", model.coef_[0])
print("Location coefficient:", model.coef_[1])
print("Distance coefficient:", model.coef_[2])

y_pred = model.predict(x)
mse= mean_squared_error(y,y_pred)
mae= mean_absolute_error(y,y_pred)
r2 = r2_score(y,y_pred)
print("MSE :",mse)
print("MAE :",mae)
print("R2 :",r2)

#Step 7 Predection
new_house=[[2100,9,4]]
prediction = model.predict(new_house)
print("Predicted Salary :", prediction[0], "Lakhs")

# Graph
fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(
    df['Area'],
    df['Location_score'],
    df['Distance'],
    s=80
)

ax.set_xlabel('Area(sqft)')
ax.set_ylabel('Location_score')
ax.set_zlabel('Distance')

ax.set_title('Area + Location_score vs Distance')
plt.show()
```
