```python
import pandas as pd



data = {
    "Age": [21, 25, 19, 32, 28, 24, 35, 22, 29, 31,
            20, 27, 38, 23, 41, 26, 30, 34, 18, 36,
            27, 33, 45, 24, 39],

    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male"],

    "Education": ["Graduate", "Postgraduate", "Undergraduate", "Graduate", "Postgraduate",
                  "Graduate", "Graduate", "Undergraduate", "Postgraduate", "Graduate",
                  "Undergraduate", "Graduate", "Postgraduate", "Undergraduate", "Graduate",
                  "Postgraduate", "Graduate", "Graduate", "Undergraduate", "Postgraduate",
                  "Graduate", "Graduate", "Postgraduate", "Undergraduate", "Graduate"],

"Review": ["Good", "Excellent", "Average", "Good", "Excellent",
               "Poor", "Good", "Average", "Good", "Excellent",
               "Poor", "Good", "Excellent", "Average", "Good",
               "Excellent", "Average", "Good", "Poor", "Excellent",
               "Good", "Average", "Excellent", "Poor", "Good"],

    "Purchased": ["Yes", "Yes", "No", "Yes", "Yes",
                  "No", "Yes", "No", "Yes", "Yes",
                  "No", "Yes", "Yes", "No", "Yes",
                  "Yes", "No", "Yes", "No", "Yes",
                  "Yes", "No", "Yes", "No", "Yes"],


}

df = pd.DataFrame(data)

print(df)



```

```python
df=df.iloc[:,2:]
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.iloc[:,0:2],df.iloc[:,-1],test_size=0.2)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.preprocessing import OrdinalEncoder

ordinal = OrdinalEncoder(
    categories=[
        ['Undergraduate', 'Graduate', 'Postgraduate'],
        ['Poor', 'Average', 'Good', 'Excellent']
    ]
)

ordinal.fit(x_train)
```

```python
x_train=ordinal.transform(x_train)
```

```python
x_train
```
