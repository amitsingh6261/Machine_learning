```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# -----------------------------------
# 1. Create Dataset
# -----------------------------------

data = {
    "Student": [
        "Aman", "Riya", "Karan", "Neha", "Rahul",
        "Priya", "Vikas", "Anjali", "Rohit", "Sneha",
        "Arjun", "Pooja", "Nikhil", "Simran", "Aditya"
    ],

    "Age": [
        19, 20, 18, 21, 19,
        20, 18, 21, 19, 20,
        18, 21, 19, 20, 18
    ],

    "Study_Level": [
        "Medium", "High", "Low", "High", "Medium",
        "Low", "High", "Medium", "Low", "High",
        "Medium", "High", "Low", "Medium", "High"
    ],

    "Attendance": [
        "Good", "Excellent", "Poor", "Excellent", "Good",
        "Average", "Excellent", "Good", "Poor", "Excellent",
        "Average", "Good", "Poor", "Average", "Excellent"
    ],

    "Performance": [
        "Good", "Excellent", "Average", "Excellent", "Good",
        "Poor", "Excellent", "Good", "Poor", "Excellent",
        "Average", "Good", "Poor", "Average", "Excellent"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

```

```python
x = df[["Study_Level", "Attendance", "Performance"]]
```

```python
ordinal = OrdinalEncoder(
    categories=[
        ["Low", "Medium", "High"],
        ["Poor", "Average", "Good", "Excellent"],
        ["Poor", "Average", "Good", "Excellent"]
    ]
)

x_encoded = ordinal.fit_transform(x)
```

```python
encoded_df = pd.DataFrame(
    x_encoded,
    columns=[
        "Study_Level_Encoded",
        "Attendance_Encoded",
        "Performance_Encoded"
    ]
)

print("\nEncoded Data:")
print(encoded_df)
```

```python
final_df = pd.concat(
    [df, encoded_df],
    axis=1
)

print("\nFinal Dataset:")
print(final_df)
```
